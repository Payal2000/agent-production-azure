## 1.1 What Is Voice Live?

**Voice Live API** enables real-time audio conversations with Foundry agents. Users speak naturally, and the agent responds with synthesized speech — all in real-time over a WebSocket connection.

### Architecture

```
┌──────────────────────────────────────────────────────────────────────────┐
│                     VOICE LIVE ARCHITECTURE                             │
│                                                                         │
│  ┌──────────┐     ┌─────────────────┐     ┌───────────────────────┐     │
│  │  User    │     │  Voice Live API  │     │ Foundry Agent Service │     │
│  │ (Mic +   │◄───▶│  (WebSocket)     │◄───▶│ (Agent + Tools)       │     │
│  │  Speaker)│     │                 │     └───────────────────────┘     │
│  └──────────┘     │  ┌───────────┐  │                                   │
│                   │  │ Azure     │  │     ┌───────────────────────┐     │
│                   │  │ Speech    │  │     │ Knowledge / Tools     │     │
│                   │  │ Services  │  │     │ (Web Search, MCP,     │     │
│                   │  │ (STT/TTS) │  │     │  File Search, etc.)   │     │
│                   │  └───────────┘  │     └───────────────────────┘     │
│                   └─────────────────┘                                   │
│                                                                         │
│  Audio: 24kHz PCM16 mono, base64-encoded                               │
│  No separate audio model deployment needed — uses agent's LLM          │
└──────────────────────────────────────────────────────────────────────────┘
```

### Key Features

| Feature | Description |
|---------|-------------|
| **Barge-in** | User can interrupt the agent mid-speech; agent stops and processes new input |
| **Semantic VAD** | Voice Activity Detection uses semantic understanding to detect end-of-turn |
| **Noise Reduction** | Filters background noise from user audio |
| **Echo Cancellation** | Prevents agent from hearing its own playback |
| **Proactive Greeting** | Agent speaks first when session starts |
| **Interim Responses** | Bridging phrases ("Let me check...") while tools execute |
| **Tool Integration** | Agent can call any attached tools during voice conversation |

## 1.2 Voice Live Configuration

Session configuration is stored in the agent's **metadata** field (512-character limit). Large configs are chunked across multiple metadata keys.

### Voice Config Parameters

| Parameter | Default | Description |
|-----------|---------|-------------|
| **voice** | `en-US-Ava:DragonHDLatestNeural` | Azure Speech TTS voice |
| **temperature** | `0.8` | Model temperature for voice responses |
| **vad_type** | `semantic_vad` | Voice Activity Detection type |
| **noise_reduction** | `true` | Filter background noise |
| **echo_cancellation** | `true` | Prevent feedback loop |
| **greeting** | (configurable) | Proactive first message from agent |
| **audio_format** | `pcm16` | 24kHz PCM16 mono |

### Event Types

| Event | Direction | Description |
|-------|-----------|-------------|
| `SESSION_UPDATED` | Server → Client | Session config confirmed |
| `SPEECH_STARTED` | Server → Client | Agent begins speaking |
| `SPEECH_STOPPED` | Server → Client | Agent finishes speaking |
| `RESPONSE_AUDIO_DELTA` | Server → Client | Chunk of audio data (base64) |
| `INPUT_AUDIO_BUFFER_APPEND` | Client → Server | User audio chunk |
| `INPUT_AUDIO_BUFFER_COMMIT` | Client → Server | Signal end of user utterance |
| `RESPONSE_DONE` | Server → Client | Full response completed |
| `CONVERSATION_ITEM_CREATED` | Server → Client | New conversation turn logged |

### Interim Response Config (Latency Bridging)

When tools take time to execute, configure interim responses so the agent says bridging phrases:

```json
{
  "interim_response_config": {
    "enabled": true,
    "instructions": "Say a brief phrase like 'Let me look that up for you' while waiting for tools."
  }
}
```

````python
# 1.3 Create an Agent with Voice Live Config
# Prerequisites: pip install azure-ai-voicelive pyaudio

print("""
import json
from azure.identity import DefaultAzureCredential
from azure.ai.projects import AIProjectClient
from azure.ai.projects.models import PromptAgentDefinition, WebSearchTool

PROJECT_ENDPOINT = "your_project_endpoint"

project = AIProjectClient(
    endpoint=PROJECT_ENDPOINT,
    credential=DefaultAzureCredential(),
)

# Voice Live session config stored in agent metadata
# 512-char limit per key — chunk if needed
voice_config = {
    "voice": "en-US-Ava:DragonHDLatestNeural",
    "temperature": 0.8,
    "vad_type": "semantic_vad",
    "noise_reduction": True,
    "echo_cancellation": True,
    "greeting": "Hello! How can I help you today?",
    "interim_response_config": {
        "enabled": True,
        "instructions": "Say a brief phrase while waiting for tools."
    },
}

# Chunk metadata if it exceeds 512 chars
config_str = json.dumps(voice_config)
metadata = {}
CHUNK_SIZE = 512
for i in range(0, len(config_str), CHUNK_SIZE):
    metadata[f"voice_config_{i // CHUNK_SIZE}"] = config_str[i:i + CHUNK_SIZE]

# Create agent with voice config in metadata
agent = project.agents.create_version(
    agent_name="voice-assistant",
    definition=PromptAgentDefinition(
        model="gpt-4.1-mini",
        instructions="You are a friendly voice assistant. Keep responses concise for speech.",
        tools=[WebSearchTool()],
        metadata=metadata,
    ),
)
print(f"Agent created: {agent.name}, Version: {agent.version}")
print(f"Metadata keys: {list(metadata.keys())}")
""")
````

````python
# 1.4 Connect to Voice Live Session and Handle Events
print("""
import asyncio
import pyaudio
from azure.ai.voicelive import VoiceLiveClient, VoiceLiveConfig
from azure.identity import DefaultAzureCredential

# Audio settings: 24kHz PCM16 mono
SAMPLE_RATE = 24000
CHANNELS = 1
FORMAT = pyaudio.paInt16
CHUNK = 1024

async def run_voice_session():
    config = VoiceLiveConfig(
        project_endpoint=PROJECT_ENDPOINT,
        credential=DefaultAzureCredential(),
        agent_name="voice-assistant",
    )
    
    client = VoiceLiveClient(config)
    
    # Initialize PyAudio
    pa = pyaudio.PyAudio()
    input_stream = pa.open(format=FORMAT, channels=CHANNELS,
                           rate=SAMPLE_RATE, input=True, frames_per_buffer=CHUNK)
    output_stream = pa.open(format=FORMAT, channels=CHANNELS,
                            rate=SAMPLE_RATE, output=True, frames_per_buffer=CHUNK)
    
    async with client.connect() as session:
        print("Voice session connected.")
        
        # Handle events from the server
        async for event in session.events():
            if event.type == "SESSION_UPDATED":
                print("Session configured successfully.")
            
            elif event.type == "SPEECH_STARTED":
                print("Agent is speaking...")
            
            elif event.type == "RESPONSE_AUDIO_DELTA":
                # Decode base64 audio and play through speaker
                import base64
                audio_bytes = base64.b64decode(event.delta)
                output_stream.write(audio_bytes)
            
            elif event.type == "SPEECH_STOPPED":
                print("Agent finished speaking.")
            
            elif event.type == "RESPONSE_DONE":
                print(f"Response complete. Tokens used: {event.usage}")
        
        # Send user audio in a separate task
        async def send_audio():
            while True:
                data = input_stream.read(CHUNK, exception_on_overflow=False)
                await session.send_audio(data)
                await asyncio.sleep(0.01)
        
        await send_audio()
    
    input_stream.close()
    output_stream.close()
    pa.terminate()

# asyncio.run(run_voice_session())
print("Voice session example ready. Uncomment asyncio.run() to execute.")
""")
````

### Comparison Table

| Feature | Web Search | Bing Grounding | Bing Custom Search |
|---------|------------|---------------|--------------------|
| **Status** | GA | GA | GA |
| **Bing resource** | Not needed | Required | Required |
| **Model support** | OpenAI models only | OpenAI + non-OpenAI | OpenAI + non-OpenAI |
| **Domain restriction** | No | No | Yes |
| **Key parameters** | `user_location`, `search_context_size` | `count`, `freshness`, `market`, `set_lang` | Same as Bing Grounding + `custom_config_id` |
| **Data boundary** | Data leaves Azure compliance boundary | Data leaves Azure compliance boundary | Data leaves Azure compliance boundary |
| **Recommended for** | Most use cases | Non-OpenAI models, existing Bing resources | Restricting search to specific websites |

> **Data Boundary Warning:** All web grounding tools send data outside the Azure compliance boundary. Sensitive data should not be included in queries sent to web search.

````python
# 2.5 Web Search Tool with Location
from azure.identity import DefaultAzureCredential
from azure.ai.projects import AIProjectClient
from azure.ai.projects.models import (
    PromptAgentDefinition,
    WebSearchTool,
)

project = AIProjectClient(
    endpoint=PROJECT_ENDPOINT,
    credential=DefaultAzureCredential(),
)
openai_client = project.get_openai_client()

# Create agent with Web Search and user location
agent = project.agents.create_version(
    agent_name="web-search-agent",
    definition=PromptAgentDefinition(
        model="gpt-4.1-mini",
        instructions="You are a helpful assistant. Use web search for current information. Always cite sources.",
        tools=[
            WebSearchTool(
                user_location={
                    "type": "approximate",
                    "city": "Seattle",
                    "state": "Washington",
                    "country": "United States",
                    "country_code": "US",
                },
                search_context_size="medium",
            )
        ],
    ),
)
print(f"Agent created: {agent.name}, Version: {agent.version}")

# Generate a response using web search
response = openai_client.responses.create(
    input="What is the weather forecast for Seattle today?",
    extra_body={"agent_reference": {"name": agent.name, "type": "agent_reference"}},
)
print(f"Response: {response.output_text[:300]}...")
````

## 3.1 What Is MCP?

**Model Context Protocol (MCP)** is a standard interface that enables agents to interact with external tools and data sources. MCP servers expose tools that agents can discover and invoke.

```
┌──────────────────────────────────────────────────────────────────┐
│                    MCP TOOL FLOW                                │
│                                                                 │
│  ┌─────────┐     ┌──────────────────┐     ┌──────────────────┐ │
│  │ Agent   │────▶│ Foundry Agent     │────▶│ MCP Server       │ │
│  │ (Model) │     │ Service           │     │ (Remote)         │ │
│  │         │◀────│ (Auth + Routing)  │◀────│                  │ │
│  └─────────┘     └──────────────────┘     │ ┌──────────────┐ │ │
│                                           │ │ Tool 1       │ │ │
│  1. Model decides to call MCP tool        │ │ Tool 2       │ │ │
│  2. Agent Service authenticates            │ │ Tool 3       │ │ │
│  3. Request forwarded to MCP server        │ └──────────────┘ │ │
│  4. MCP server executes and returns        └──────────────────┘ │
│  5. Result fed back to model                                    │
└──────────────────────────────────────────────────────────────────┘
```

### MCP Tool Configuration Parameters

| Parameter | Required | Description |
|-----------|---------|-------------|
| `server_url` | Yes | URL of the remote MCP server |
| `server_label` | Yes | Display name for the server |
| `require_approval` | No | `always`, `never`, or per-tool list |
| `allowed_tools` | No | Restrict which tools the agent can call |
| `project_connection_id` | No | Connection for authentication |

### require_approval Options

| Value | Behavior |
|-------|----------|
| `"never"` | Agent calls tools without user approval |
| `"always"` | Every tool call requires user approval |
| `{"always": {"tool_names": ["delete_record"]}}` | Only specified tools require approval |
| `{"never": {"tool_names": ["read_data"]}}` | Only specified tools skip approval |

## 3.2 MCP Authentication Methods

| Method | Identity | Key Characteristics |
|--------|----------|-------------------|
| **Key-based** | API key / PAT | Stored in project connection; shared across agents |
| **Entra (Agent Identity)** | Per-agent (published) or shared (dev) | Token exchange via agent blueprint; no secrets to manage |
| **Entra (Project Managed Identity)** | All agents share project MI | System-assigned managed identity; cannot distinguish agents |
| **OAuth Identity Passthrough** | Per-user | User consents; agent acts on user's behalf; best for personal data |
| **Unauthenticated** | None | No auth required; for public MCP servers |

### MCP Approval Flow

When `require_approval` is set and a tool call needs approval, the response includes an `mcp_approval_request` output item:

```
┌───────────────────────────────────────────────────────────┐
│                MCP APPROVAL FLOW                         │
│                                                          │
│  1. Agent generates response with tool call              │
│  2. Response includes mcp_approval_request               │
│     ├── tool_name: "delete_record"                       │
│     ├── arguments: {"id": "123"}                         │
│     └── approval_request_id: "req_abc"                   │
│  3. Your app shows approval UI to user                   │
│  4. User approves or denies                              │
│  5. Your app sends approval decision back                │
│     ├── type: mcp_approval_response                      │
│     ├── approval_request_id: "req_abc"                   │
│     └── approve: true/false                              │
│  6. If approved, tool executes; if denied, agent adapts  │
└───────────────────────────────────────────────────────────┘
```

## 3.3 Hosting MCP Servers


### Registering in Organizational Tool Catalog (Azure API Center)

## 4.1 What Is A2A?

**Agent-to-Agent (A2A)** enables cross-agent communication via the [A2A protocol](https://a2a-protocol.org). An agent can delegate tasks to other agents — even agents hosted on different platforms.

### A2A Tool vs. Multi-Agent Workflow

```
┌──────────────────────────────────────────────────────────────────────┐
│          A2A TOOL vs MULTI-AGENT WORKFLOW                           │
│                                                                     │
│  A2A TOOL (Caller keeps control)                                    │
│  ┌──────────┐     ┌──────────┐                                      │
│  │ Agent A  │────▶│ Agent B  │  A calls B as a tool.                │
│  │ (Caller) │◀────│ (Callee) │  A gets B's result.                  │
│  │ Controls │     │ Executes │  A continues processing.             │
│  │ flow     │     │ task     │  A owns the conversation.            │
│  └──────────┘     └──────────┘                                      │
│                                                                     │
│  MULTI-AGENT WORKFLOW (Callee takes over)                           │
│  ┌──────────┐     ┌──────────┐                                      │
│  │ Agent A  │────▶│ Agent B  │  A hands off entirely.               │
│  │ (Router) │     │ (Owner)  │  B takes full control.               │
│  │ Hands    │     │ Completes│  B responds directly to user.        │
│  │ off      │     │ task     │  A is out of the loop.               │
│  └──────────┘     └──────────┘                                      │
│                                                                     │
│  Use A2A Tool when: caller needs result to continue processing      │
│  Use Workflow when: callee should own the rest of the conversation   │
└──────────────────────────────────────────────────────────────────────┘
```

### Authentication Options

| Method | Description |
|--------|-------------|
| **Key-based** | API key stored in project connection |
| **OAuth** | OAuth 2.0 client credentials |
| **Managed Identity** | Azure managed identity for Azure-hosted agents |
| **Agent Identity** | Entra agent identity (per-agent after publish) |

### Hosting A2A Endpoints

| Option | How |
|--------|-----|
| **Foundry Agent** | Enable A2A protocol on any Foundry agent — it becomes an A2A endpoint |
| **Control Plane** | Register external A2A endpoints in Foundry's agent registry |
| **Custom Build** | Build your own A2A server following the A2A protocol spec |

## 5.1 Queue-Based Tool Integration

The Azure Functions tool uses a **queue-based pattern** for asynchronous tool execution:

```
┌──────────────────────────────────────────────────────────────────────────┐
│                 AZURE FUNCTIONS TOOL — QUEUE PATTERN                    │
│                                                                         │
│  ┌─────────┐    ┌───────────────┐    ┌──────────────┐    ┌───────────┐ │
│  │ Agent   │───▶│ Input Queue   │───▶│ Azure        │───▶│ Output    │ │
│  │ Service │    │ (Storage)     │    │ Function     │    │ Queue     │ │
│  │         │◀───│               │    │ (Your code)  │    │ (Storage) │ │
│  └─────────┘    └───────────────┘    └──────────────┘    └───────────┘ │
│       │                                                       │        │
│       └──────── Polls output queue using CorrelationId ───────┘        │
│                                                                         │
│  1. Agent places message in INPUT queue (with CorrelationId)           │
│  2. Azure Function triggers on queue message                            │
│  3. Function processes and places result in OUTPUT queue                │
│  4. Agent polls OUTPUT queue using CorrelationId to find result         │
└──────────────────────────────────────────────────────────────────────────┘
```

### When to Use Azure Functions vs Function Calling

| Feature | Azure Functions Tool | Function Calling |
|---------|---------------------|------------------|
| **Execution** | Azure Functions executes asynchronously | Your app executes locally |
| **Pattern** | Queue-based (fire and poll) | Synchronous round-trip |
| **Best for** | Long-running tasks, background processing | Quick computations, API calls |
| **Infrastructure** | Azure Functions + Storage Queues | Your app process |
| **Scaling** | Auto-scales with Azure Functions | Scales with your app |

### Integration Options: MCP Servers vs Queue-Based Tools

| Feature | MCP via Azure Functions | Queue-Based Azure Functions Tool |
|---------|------------------------|----------------------------------|
| **Protocol** | MCP (HTTP-based) | Storage Queue messages |
| **Communication** | Synchronous HTTP | Asynchronous queues |
| **Setup** | `azd` template, streamable HTTP | Input/output queue bindings |
| **Best for** | Real-time tool calls | Long-running background tasks |

### Flex Consumption Plan Benefits

| Benefit | Description |
|---------|-------------|
| Scale to zero | No cost when idle |
| Auto-scale | Handles burst traffic automatically |
| Pay-per-use | Only pay for execution time |
| VNet integration | Secure network access |

````python
# 5.2 Azure Functions Tool with Storage Queue Bindings
print("""
from azure.identity import DefaultAzureCredential
from azure.ai.projects import AIProjectClient
from azure.ai.projects.models import (
    PromptAgentDefinition,
    AzureFunctionTool,
    AzureFunctionToolDefinition,
    AzureFunctionBinding,
    AzureFunctionStorageQueue,
)

project = AIProjectClient(
    endpoint=PROJECT_ENDPOINT,
    credential=DefaultAzureCredential(),
)
openai_client = project.get_openai_client()

STORAGE_CONN_STRING = "your_storage_connection_string"

agent = project.agents.create_version(
    agent_name="azure-func-agent",
    definition=PromptAgentDefinition(
        model="gpt-4.1-mini",
        instructions="You process orders using the order processing function.",
        tools=[
            AzureFunctionTool(
                azure_function=AzureFunctionToolDefinition(
                    name="process_order",
                    description="Process a customer order and return order status.",
                    parameters={
                        "type": "object",
                        "properties": {
                            "order_id": {
                                "type": "string",
                                "description": "The order ID to process",
                            },
                            "action": {
                                "type": "string",
                                "enum": ["confirm", "cancel", "status"],
                                "description": "Action to perform on the order",
                            },
                        },
                        "required": ["order_id", "action"],
                    },
                    input_binding=AzureFunctionBinding(
                        storage_queue=AzureFunctionStorageQueue(
                            queue_service_uri=f"{STORAGE_CONN_STRING}/orders-input",
                            queue_name="orders-input",
                        )
                    ),
                    output_binding=AzureFunctionBinding(
                        storage_queue=AzureFunctionStorageQueue(
                            queue_service_uri=f"{STORAGE_CONN_STRING}/orders-output",
                            queue_name="orders-output",
                        )
                    ),
                )
            )
        ],
    ),
)
print(f"Azure Functions agent: {agent.name}")
""")
````

````python
# 5.3 Queue-Triggered Azure Function (the function that processes messages)
print("""
# This is the Azure Function code (function_app.py) that processes queue messages.
# Deploy this as an Azure Function with queue trigger.

import azure.functions as func
import json
import logging

app = func.FunctionApp()

@app.queue_trigger(
    arg_name="msg",
    queue_name="orders-input",
    connection="AzureWebJobsStorage",
)
@app.queue_output(
    arg_name="outputMsg",
    queue_name="orders-output",
    connection="AzureWebJobsStorage",
)
def process_order(msg: func.QueueMessage, outputMsg: func.Out[str]):
    \"\"\"Process an order from the input queue and place result in output queue.\"\"\"
    
    body = json.loads(msg.get_body().decode("utf-8"))
    correlation_id = body.get("CorrelationId")  # Must preserve this!
    arguments = json.loads(body.get("Value", "{}"))
    
    order_id = arguments.get("order_id", "unknown")
    action = arguments.get("action", "status")
    
    logging.info(f"Processing order {order_id}, action: {action}")
    
    # Your business logic here
    result = {
        "order_id": order_id,
        "action": action,
        "status": "confirmed" if action == "confirm" else "processed",
        "message": f"Order {order_id} has been {action}ed successfully.",
    }
    
    # Output must include CorrelationId for the agent to match the result
    output = {
        "Value": json.dumps(result),
        "CorrelationId": correlation_id,  # Critical: agent uses this to find result
    }
    outputMsg.set(json.dumps(output))
    logging.info(f"Result placed in output queue for correlation {correlation_id}")
""")
````

```
│                                                                      │
│  ┌────────────┐    ┌──────────────┐    ┌──────────────────┐         │
```

### Safety Checks

### pending_safety_checks Pattern

```
┌───────────────────────────────────────────────────────────────┐
│            SAFETY CHECK FLOW                                  │
│                                                               │
│  1. Agent generates computer_call with pending_safety_checks  │
│     └── [{"id": "sc_1", "code": "sensitive_domain",          │

│                                                               │
│  2. Your app shows warning to user                            │
│                                                               │
│  3. If user approves, resend with acknowledged_safety_checks  │
│     └── [{"id": "sc_1"}]                                     │
│                                                               │
│  4. If user denies, stop execution                            │
│                                                               │
│  NEVER auto-acknowledge safety checks without user consent!   │
└───────────────────────────────────────────────────────────────┘
```

````python
# 6.5 Computer Use Tool with Safety Checks
print("""
from azure.identity import DefaultAzureCredential
from azure.ai.projects import AIProjectClient
from azure.ai.projects.models import (
    PromptAgentDefinition,
    ComputerUsePreviewTool,
)

project = AIProjectClient(
    endpoint=PROJECT_ENDPOINT,
    credential=DefaultAzureCredential(),
)
openai_client = project.get_openai_client()

# Computer Use requires the computer-use-preview model
agent = project.agents.create_version(
    agent_name="computer-use-agent",
    definition=PromptAgentDefinition(
        model="computer-use-preview",  # Must use this specific model
        instructions="You can interact with the computer to complete tasks.",
        tools=[
            ComputerUsePreviewTool(
                display_width=1920,
                display_height=1080,
                environment="browser",  # or "desktop"
            )
        ],
    ),
)
print(f"Computer Use agent: {agent.name}")

# Handle the screenshot-action loop with safety checks
import base64

response = openai_client.responses.create(
    input="Open the calculator app and compute 1234 * 5678",
    extra_body={"agent_reference": {"name": agent.name, "type": "agent_reference"}},
)

while True:
    computer_calls = [item for item in response.output if item.type == "computer_call"]
    if not computer_calls:
        break  # No more actions needed
    
    for call in computer_calls:
        # Check for pending safety checks
        if call.pending_safety_checks:
            for check in call.pending_safety_checks:
                print(f"Safety warning: [{check.code}] {check.message}")
            
            # In production, show to user and get consent
            user_approved = True  # Prompt user in real app
            
            if not user_approved:
                print("User denied. Stopping.")
                break
            
            acknowledged = [{"id": c.id} for c in call.pending_safety_checks]
        else:
            acknowledged = []
        
        # Execute the action (click, type, scroll, etc.)
        action = call.action
        print(f"Executing: {action.type} at ({getattr(action, 'x', '-')}, {getattr(action, 'y', '-')})")
        
        # Take a screenshot after executing the action
        # In real implementation, capture from the VM/environment
        screenshot_base64 = "<base64-encoded-screenshot>"
        
        # Send screenshot back to continue the loop
        response = openai_client.responses.create(
            input=[{
                "type": "computer_call_output",
                "call_id": call.call_id,
                "output": {
                    "type": "computer_screenshot",
                    "image_url": f"data:image/png;base64,{screenshot_base64}",
                },
                "acknowledged_safety_checks": acknowledged,
            }],
            extra_body={"agent_reference": {"name": agent.name, "type": "agent_reference"}},
        )

# Final text output
print(f"Result: {response.output_text}")
""")
````

## 9.3 OAuth Identity Passthrough — Consent Flow

When using OAuth identity passthrough, the user must consent before the agent can act on their behalf:

```
┌───────────────────────────────────────────────────────────────────┐
│            OAUTH CONSENT FLOW                                    │
│                                                                  │
│  1. User sends message to agent                                  │
│  2. Agent tries to call MCP tool requiring user identity         │
│  3. Response includes oauth_consent_request:                     │
│     ├── consent_link: "https://login.microsoftonline.com/..."    │
│     ├── server_label: "outlook-tools"                            │
│     └── scopes: ["Mail.Read", "Calendar.ReadWrite"]              │
│  4. Your app presents consent_link to user                       │
│  5. User clicks link, authenticates, grants consent              │
│  6. Your app resends the original request                        │
│  7. Agent now has user's token and calls MCP tool                │
│  8. MCP server receives user's access token                      │
└───────────────────────────────────────────────────────────────────┘
```

# PART 10: INTERVIEW Q&A — KEY CONCEPTS
---

### Q1: What is Voice Live and how does barge-in work?
**A:** Voice Live API enables real-time audio conversations with Foundry agents over WebSocket connections. Audio is 24kHz PCM16 mono, base64-encoded. **Barge-in** means the user can interrupt the agent mid-speech — when the system detects user speech (via Semantic VAD), it immediately stops the agent's audio output and begins processing the new user input. This creates natural conversational flow. The session config is stored in agent metadata (512-char limit per key, chunked if needed).

---

### Q2: Compare Web Search vs Bing Grounding vs Bing Custom Search.
**A:** (1) **Web Search** (recommended) — GA, no Bing resource needed, supports `user_location` and `search_context_size` (low/medium/high), but only works with OpenAI models. (2) **Bing Grounding** — requires your own Bing resource, supports non-OpenAI models, offers `count`, `freshness`, `market`, and `set_lang` parameters. (3) **Bing Custom Search** — same as Bing Grounding but restricts search to specific domains via a `custom_config_id` configured in Bing Custom Search portal. All three send data outside the Azure compliance boundary.

---

### Q3: How does MCP authentication work? Compare the 5 methods.
**A:** (1) **Key-based** — API key/PAT stored in project connection; shared across agents; simplest. (2) **Entra Agent Identity** — per-agent identity after publishing (shared in dev); no secrets to manage; 4-stage token exchange. (3) **Entra Project Managed Identity** — system-assigned MI shared by all agents; cannot distinguish between agents at the MCP server. (4) **OAuth Identity Passthrough** — per-user consent; agent acts on behalf of user; best for personal data (email, calendar); requires consent flow with `oauth_consent_request`. (5) **Unauthenticated** — no auth; for public MCP servers.

---

### Q4: What is the A2A tool vs workflow? When to use each?
**A:** **A2A Tool** — the caller agent invokes another agent as a tool, gets the result back, and continues processing. The caller keeps control of the conversation. **Multi-Agent Workflow** — the caller hands off entirely to the callee, which takes full ownership. Use A2A Tool when the caller needs the result to continue (e.g., coordinator agent gathering data from specialist agents). Use Workflow when the callee should own the rest of the conversation (e.g., routing to a specialist who handles everything).

---

### Q5: How does the Azure Functions tool work with queues?
**A:** The Azure Functions tool uses a queue-based asynchronous pattern: (1) Agent places a message in an **input Storage Queue** with a `CorrelationId`. (2) Azure Function triggers on the queue message, processes it. (3) Function places the result in an **output Storage Queue** with the same `CorrelationId`. (4) Agent polls the output queue using the `CorrelationId` to find and retrieve the result. This is different from function calling (synchronous, your app executes) and MCP (synchronous HTTP). Use it for long-running background tasks.

---

### Q6: Compare browser automation vs computer use tool.
**A:** **Browser Automation** — Playwright-based, DOM-level selectors, works with any supported model, faster and more precise for web tasks, sandboxed browser session. **Computer Use** — screenshot-based pixel-coordinate actions, requires `computer-use-preview` model specifically, works on any UI (desktop apps, not just web), slower due to screenshot analysis loop, includes safety checks (`pending_safety_checks` for `malicious_instructions`, `irrelevant_domain`, `sensitive_domain`). Use Browser Automation for web scraping/forms. Use Computer Use for desktop apps or cross-application workflows.

---

### Q7: What is Foundry IQ and how does agentic retrieval differ from standard RAG?
**A:** Foundry IQ is a managed knowledge base service that combines knowledge sources (Blob, SharePoint, OneLake, Web) with an agentic retrieval pipeline. **Standard RAG** sends a single query to a vector index. **Agentic retrieval** decomposes complex queries into focused subqueries, runs them in parallel, applies semantic reranking, and enforces permissions (ACL sync, Purview labels). It also supports configurable reasoning effort (minimal, low, medium). Foundry IQ is permission-aware — query-time enforcement ensures users only see content they have access to.

---

### Q8: What is the Custom Code Interpreter and when to use it?
**A:** Custom Code Interpreter extends the standard Code Interpreter via an MCP-based tool backed by **Azure Container Apps Dynamic Sessions**. Use it when you need: (1) custom Python packages not in the standard sandbox, (2) custom container images with specialized environments, (3) dedicated compute resources, or (4) configurable network access (standard Code Interpreter has no outbound access). It requires more setup (Container Apps session pool) but gives full control over the execution environment.

---

### Q9: How does SharePoint tool maintain security with identity passthrough?
**A:** The SharePoint tool uses **identity passthrough (On-Behalf-Of)** — the agent accesses SharePoint content using the calling user's Microsoft 365 permissions. It uses the Microsoft 365 Copilot Retrieval API under the hood. This means: (1) users only see documents they have permission to access, (2) no service-level credentials that bypass user permissions, (3) audit trails show the user's identity. Requires either an M365 Copilot license or pay-as-you-go metering.

---

### Q10: What are pending_safety_checks in computer use?
**A:** When the `computer-use-preview` model detects potentially dangerous actions, it includes `pending_safety_checks` in the `computer_call` output. Each check has an `id`, `code` (one of `malicious_instructions`, `irrelevant_domain`, `sensitive_domain`), and a `message` explaining the risk. Your app must: (1) display the warning to the user, (2) get explicit user consent, (3) include `acknowledged_safety_checks` with the matching IDs in the follow-up request. Never auto-acknowledge safety checks without user consent.

---

### Q11: How do you build and register your own MCP server?
**A:** (1) Scaffold using `azd init --template azure-functions-mcp-server` for Azure Functions, or build a container for Azure Container Apps. (2) Implement tools following the MCP protocol spec (streamable HTTP). (3) Deploy to Azure. (4) Register in the organizational tool catalog by creating an Azure API Center resource, registering the MCP server as an API with environment and deployment, and granting developers Azure API Center Data Reader role. (5) Developers discover the server in Foundry Portal under Build > Tools.

---

### Q12: What is OAuth identity passthrough consent flow?
**A:** When an agent needs to call an MCP tool using the user's identity: (1) Agent attempts the tool call. (2) The response includes an `oauth_consent_request` with a `consent_link`, `server_label`, and `scopes`. (3) Your app presents the consent link to the user. (4) User clicks, authenticates with the OAuth provider, and grants the requested scopes. (5) Your app resends the original request. (6) The agent now has the user's token and can call the MCP tool. This works for both managed OAuth (Entra ID) and custom OAuth providers.

---

### Q13: What is Deep Research and how does it differ from Web Search?
**A:** Deep Research uses the `o3-deep-research` model to perform multi-step web research. Unlike regular Web Search (single query, quick results), Deep Research: (1) autonomously plans and executes multiple search iterations, (2) synthesizes findings across many sources, (3) produces comprehensive research reports with citations. It takes longer (minutes vs seconds) but delivers much deeper analysis. You do NOT attach a WebSearchTool — the model uses web search internally. Best for complex research questions requiring multi-source synthesis.

---

### Q14: How does Image Generation work in Foundry agents?
**A:** Image Generation uses the `gpt-image-1` model. You attach an `ImageGenTool` to the agent, and set the `x-ms-oai-image-generation-deployment` header to your deployment name. Parameters include `size` (1024x1024, 1024x1536, 1536x1024), `quality` (low/medium/high), `background` (transparent/opaque), `output_format` (png/jpeg/webp), and `partial_images` for progressive generation. The generated image appears in the response output as an image content block.

---

### Q15: How does the MCP approval flow work?
**A:** When `require_approval` is configured (either `"always"` or per-tool list): (1) The agent generates a response containing an `mcp_approval_request` output item with `tool_name`, `arguments`, and `approval_request_id`. (2) Your app extracts this and presents it to the user ("Agent wants to call delete_record with id=123. Allow?"). (3) User approves or denies. (4) Your app sends an `mcp_approval_response` input with the `approval_request_id` and `approve: true/false`. (5) If approved, the tool executes; if denied, the agent generates an alternative response.

---

### Q16: What are the hosting options for A2A endpoints?
**A:** Three options: (1) **Enable A2A on a Foundry agent** — any Foundry agent can expose the A2A protocol endpoint, making it callable by other agents. (2) **Register in Control Plane** — register external A2A endpoints in Foundry's agent registry so they are discoverable. (3) **Custom Build** — build your own A2A-compatible server following the a2a-protocol.org specification and host anywhere. Authentication supports key-based, OAuth, managed identity, and agent identity.

---

### Q17: What is the data boundary warning for web grounding tools?
**A:** All web grounding tools (Web Search, Bing Grounding, Bing Custom Search) send data outside the Azure compliance boundary. This means: (1) Query data flows to Bing/web search infrastructure outside your Azure region. (2) This may violate data residency requirements for regulated industries. (3) Administrators can block web search subscription-wide using `az feature register --namespace Microsoft.CognitiveServices --name BlockWebSearchInAgentService`. (4) Sensitive data should never be included in queries sent to web search.

---

### Q18: How does Foundry IQ handle permissions across different knowledge sources?
**A:** Foundry IQ is permission-aware through multiple mechanisms: (1) **ACL sync** — Access Control Lists from source systems (SharePoint, Blob) are synchronized to the search index. (2) **Microsoft Purview labels** — sensitivity labels are preserved and enforced. (3) **Query-time enforcement** — permissions are checked when the query runs, not just at index time, ensuring up-to-date access control. For SharePoint sources, the user's M365 identity determines what content they can see.

---

### Q19: What is the CorrelationId pattern in Azure Functions Tool?
**A:** The CorrelationId is a unique identifier that links an agent's tool request to the corresponding function result. The flow: (1) Agent places a message in the input queue containing a `CorrelationId`. (2) The Azure Function reads the message, processes it, and places the result in the output queue with the **same** `CorrelationId` in the output message. (3) The agent polls the output queue and uses the `CorrelationId` to find and match its specific result. Without preserving the CorrelationId in the output, the agent cannot find the function's response.

---

### Q20: What are Agent 365 MCP servers and who can use them?
**A:** Agent 365 MCP servers are pre-built MCP servers provided by Microsoft for Microsoft 365 services — Outlook, Calendar, Teams, SharePoint, Word, Excel, PowerPoint, OneDrive. They are currently available for **Frontier tenants**. They use OAuth identity passthrough, so the agent acts with the individual user's M365 permissions. This enables agents to read emails, create calendar events, send Teams messages, edit documents, and more — all while respecting the user's existing M365 access controls.

---

### Q21: When would you use the OpenAPI tool vs MCP tool vs Function Calling?
**A:** **OpenAPI tool** — when you have an existing REST API with an OpenAPI 3.0/3.1 spec; supports anonymous, API key, or managed identity auth; the Foundry service calls the API directly. **MCP tool** — when connecting to an MCP-compatible server that exposes multiple tools; supports 5 auth methods including user identity passthrough; best for tool discovery and rich tool ecosystems. **Function Calling** — when you need custom logic executed in your own application process; your app receives the function call request, executes locally, and returns the result. Use Function Calling for proprietary logic, MCP for shared tool servers, and OpenAPI for existing REST APIs.

---

### Q22: How do interim responses work in Voice Live?
**A:** Interim responses are bridging phrases the agent speaks while waiting for slow tool execution. When configured with `interim_response_config: {enabled: true, instructions: "..."}`, the agent generates short phrases like "Let me look that up for you" or "One moment please" while tools are executing. This prevents awkward silence during voice conversations. The instructions field guides the tone and content of these bridging phrases. Once the tool returns, the agent speaks the actual response.
