## 1.2 Two Categories of Foundry Models

The model catalog is organized into two main categories:

### Models Sold by Azure ("Azure Direct")

Hosted and sold by Microsoft under Microsoft Product Terms. Key characteristics:

- **Microsoft support** included with enterprise-grade SLAs
- **Deep Azure integration** with services and infrastructure
- **Responsible AI** — subject to internal review based on Microsoft's standards
- **Transparency reports** provide visibility to model risks and mitigations
- **Fungible provisioned throughput** — flexibly share quota across models
- **Billing** — via Azure meters as First Party Consumption Services

### Models from Partners and Community

Provided by third-party organizations, partners, research labs, and community contributors:

- **Developed and supported** by external providers
- **Diverse and specialized** — niche or broad use cases
- **Community-driven innovation** — rapid availability of cutting-edge models
- **Billing** — through Azure Marketplace
- Deployable using **managed compute** or **serverless deployment** options

### How to Choose

| Criteria | Models Sold by Azure | Models from Partners |
|---|---|---|
| **Support** | Microsoft-provided, enterprise SLAs | Provider-supported, varying SLAs |
| **Integration** | Deep Azure integration | Standard Azure AI integration |
| **Best for** | Enterprise production, compliance | Specialized/innovation use cases |
| **Innovation** | Stable, validated releases | Rapid access to cutting-edge models |
| **Billing** | Azure meters (First Party) | Azure Marketplace (Third Party) |

> **Note:** For all models, customers remain responsible for complying with the law, reviewing model documentation, selecting appropriate models, and implementing content safety measures.

## 4.1 Serverless Deployments vs. Managed Compute

| Feature | Managed Compute | Serverless Deployment |
|---|---|---|
| **Hosting** | Dedicated VMs in your subscription | Microsoft-managed infrastructure |
| **Billing** | VM core hours | Pay-per-token (input/output) |
| **Authentication** | Keys + Microsoft Entra | Keys + Microsoft Entra |
| **Content Safety** | Separate Azure AI Content Safety APIs | Integrated content safety filters |
| **Network Isolation** | Managed network configuration | Follows hub PNA flag setting |
| **Best For** | Custom models, specific VM needs | API-based access, cost optimization |

> **Important:** Managed compute models require a **hub-based project** in the Foundry portal (classic).

## 4.2 Serverless Deployment Types

| Deployment Type | Processing | Billing | Best For |
|---|---|---|---|
| **Global Standard** | Routes globally | Pay-per-token | Maximum throughput |
| **Global Provisioned** | Routes globally | Reserved capacity | Consistent latency at scale |
| **Global Batch** | Routes globally | Discounted batch | Large offline workloads |
| **Data Zone Standard** | Within data zone | Pay-per-token | Data residency + throughput |
| **Data Zone Provisioned** | Within data zone | Reserved capacity | Compliance + performance |
| **Data Zone Batch** | Within data zone | Discounted batch | Compliant batch processing |
| **Standard** | Regional | Pay-per-token | Single-region workloads |
| **Regional Provisioned** | Regional | Reserved capacity | Strict region requirements |
| **Developer** | — | Free/limited | Testing and prototyping |

# PART 5: MODEL CHOICE GUIDE — GPT-5 vs GPT-4.1
---

## 5.1 When to Use GPT-5 vs GPT-4.1

GPT-5 introduces **four adjustable thinking levels**, controlling how much time and tokens the model spends reasoning. GPT-4.1 is optimized for **speed and throughput**.

### Head-to-Head Comparison

| Feature | GPT-5 | GPT-4.1 |
|---|---|---|
| **Model Type** | Reasoning | Non-reasoning, fast response |
| **Best For** | Complex reasoning, multi-hop logic | Real-time chat, high-throughput |
| **Latency** | Higher (deeper reasoning) | Lower (optimized for speed) |
| **Throughput** | Moderate | High |
| **Token Length** | 272K in / 128K out (400K total) | 128K-1M (long context) |
| **Perspective** | Structured, analytical, step-by-step | Concise, fast, conversational |
| **Variants** | GPT-5, GPT-5-mini, GPT-5-nano | GPT-4.1, GPT-4.1-mini, GPT-4.1-nano |

### GPT-5 Thinking Levels

| Level | Depth | Latency | Cost | Accuracy | Use Cases |
|---|---|---|---|---|---|
| **Minimal** | Very shallow | Fastest | Lowest | Lowest on complex tasks | Bulk operations, simple transforms |
| **Low** | Light | Fast | Low | Moderate | Triage, short answers, simple edits |
| **Medium** (default) | Moderate | Moderate | Medium | Good for most tasks | Content drafting, coding, RAG Q&A |
| **High** | Deep, multistep | Slowest | Highest | Highest | Complex planning, analysis, reasoning |

> **Note:** Parallel tool calls are NOT supported at Minimal reasoning_effort. Use Low, Medium, or High.

### Use GPT-5 When:
- Deep, multistep reasoning for planning, analysis, synthesis
- Reliability over raw speed (higher quality, fewer mistakes)
- Agentic workflows for Copilot-style tools with multi-tool orchestration
- **Examples:** Legal/financial analysis, troubleshooting, enterprise Copilots, research synthesis

### Use GPT-4.1 When:
- Low latency for real-time interactions or chatbots
- High throughput for large-scale deployments
- Long-context handling (up to 1M tokens)
- Short, factual responses
- **Examples:** Customer support bots, recommendation engines, summarization pipelines

### Latency Comparison

| Metric | GPT-5 | GPT-4.1 |
|---|---|---|
| **TTFT** | Higher (deeper model layers) | Lower |
| **TBT** | Moderate to high | Low |
| **User Perception** | May feel slower for short prompts | Snappy and responsive |

> **Tip:** Use **Provisioned Throughput** deployment for GPT-5 if you need consistent latency SLAs.

### Model Router

If unsure which model to choose, try [Model Router](https://ai.azure.com/catalog/models/model-router) — a ready-to-use solution that automatically routes requests to the optimal model, saving up to **60% on inference cost** with comparable quality.

````python
# ============================================================
# Compare GPT-5 vs GPT-4.1 — Simple Latency & Quality Test
# ============================================================

import os
import time
from azure.ai.projects import AIProjectClient
from azure.identity import DefaultAzureCredential

endpoint = os.environ["FOUNDRY_PROJECT_ENDPOINT"]

test_prompt = "Explain the trade-offs between microservices and monolithic architecture in 3 sentences."

# Models to compare (adjust to your deployed model names)
models_to_test = ["gpt-4o-mini", "gpt-4o"]

with (
    DefaultAzureCredential() as credential,
    AIProjectClient(endpoint=endpoint, credential=credential) as project_client,
):
    with project_client.get_openai_client() as openai_client:
        for model_name in models_to_test:
            try:
                start = time.time()
                response = openai_client.chat.completions.create(
                    model=model_name,
                    messages=[{"role": "user", "content": test_prompt}],
                    max_tokens=200,
                )
                elapsed = time.time() - start

                tokens_used = response.usage.total_tokens
                output = response.choices[0].message.content[:150]

                print(f"\n{'='*60}")
                print(f"Model: {model_name}")
                print(f"Latency: {elapsed:.2f}s")
                print(f"Tokens: {tokens_used}")
                print(f"Response: {output}...")
            except Exception as e:
                print(f"\n{model_name}: {e}")
````

Metrics are aggregated over 14 days (24 trials/day, 2 requests/trial at 1-hour intervals):

| Metric | Description |
|---|---|

| **Latency P50/P90/P95/P99** | Percentile latencies |
| **TTFT** | Time to first token (streaming) |
| **GTPS** | Generated tokens per second |
| **TTPS** | Total tokens per second (input + output) |
| **Time Between Tokens** | Inter-token latency |

Default test: 800 input / 200 output tokens, streaming enabled, East US region.

### Cost Benchmarks

Measure **actual cost to execute** each model on quality benchmark datasets:
- Uses real token counts (input, reasoning, output)
- Reflects model-specific reasoning effort configuration
- Reported in **USD per benchmark run**
- Lower = more cost-efficient

## 11.3 Cost Optimization Tips

1. **Right-size your model** — don't use GPT-5 for simple Q&A; GPT-4.1-nano may suffice
2. **Use reasoning levels wisely** — Minimal/Low for simple tasks, Medium for general, High only when needed
3. **Model Router** — automatically routes to the optimal model, saving up to 60%
4. **Reduce embedding dimensions** — `text-embedding-3-small` with `dimensions=512` may be enough
5. **Batch processing** — use Global/Data Zone Batch deployments for large offline workloads
6. **Provisioned throughput** — better economics at scale vs. pay-per-token
7. **Fine-tune smaller models** — a fine-tuned GPT-4.1-nano can outperform a generic larger model
8. **Global training** — more affordable fine-tuning per token (no data residency)
